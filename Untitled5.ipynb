```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

class Consumer:
    """Represents an individual consumer with a specific willingness-to-pay (WTP)."""
    def __init__(self, consumer_id, willingness_to_pay):
        self.consumer_id = consumer_id
        self.wtp = willingness_to_pay

class PricingEngine:
    """Simulates market dynamics under different economic pricing regimes."""
    def __init__(self, baseline_wtp_array):
        self.consumers = [Consumer(i, wtp) for i, wtp in enumerate(baseline_wtp_array)]

    def simulate_lockean_flat_market(self, standard_market_price):
        """Simulates a transparent market with a singular 'Common Measure' price."""
        total_revenue = 0.0
        total_consumer_surplus = 0.0
        units_sold = 0

        for consumer in self.consumers:
            if consumer.wtp >= standard_market_price:
                total_revenue += standard_market_price
                total_consumer_surplus += (consumer.wtp - standard_market_price)
                units_sold += 1

        return {
            "Regime": "Lockean Transparent Flat Price",
            "Market Price": standard_market_price,
            "Units Sold": units_sold,
            "Corporate Revenue": round(total_revenue, 2),
            "Consumer Surplus": round(total_consumer_surplus, 2),
            "Deadweight Loss": round(sum([c.wtp for c in self.consumers if c.wtp < standard_market_price]), 2)
        }

    def simulate_algorithmic_extraction(self):
        """Simulates perfect dynamic pricing tracking real-time maximum willingness-to-pay."""
        total_revenue = 0.0
        total_consumer_surplus = 0.0
        units_sold = 0

        for consumer in self.consumers:
            extracted_price = consumer.wtp
            total_revenue += extracted_price
            total_consumer_surplus += 0.0
            units_sold += 1

        return {
            "Regime": "Algorithmic WTP Extraction",
            "Market Price": "Dynamic / Variable",
            "Units Sold": units_sold,
            "Corporate Revenue": round(total_revenue, 2),
            "Consumer Surplus": round(total_consumer_surplus, 2),
            "Deadweight Loss": 0.0
        }

# --- Empirical Execution & Visualization Block ---
if __name__ == "__main__":
    # Generate distribution mapping 1000 consumer budgets
    np.random.seed(42)
    empirical_wtp_profile = np.clip(np.random.normal(loc=350, scale=100, size=1000), 150, 700)
    wtp_sorted = np.sort(empirical_wtp_profile)[::-1]

    # Initialize engine & simulate regimes
    engine = PricingEngine(empirical_wtp_profile)
    flat_price_target = 350.00

    flat_results = engine.simulate_lockean_flat_market(flat_price_target)
    algorithmic_results = engine.simulate_algorithmic_extraction()

    # Print data summary table
    summary_df = pd.DataFrame([flat_results, algorithmic_results])
    print("\n=== SIMULATION DATA ANALYSIS OUTPUT ===\n")
    print(summary_df.to_string(index=False))
    print("\n========================================\n")

    # Generate visual asset chart
    plt.figure(figsize=(10, 6))
    plt.fill_between(range(len(wtp_sorted)), wtp_sorted, flat_price_target, where=(wtp_sorted >= flat_price_target),
                     color='#3498db', alpha=0.4, label='Extracted Consumer Surplus')
    plt.fill_between(range(len(wtp_sorted)), flat_price_target, 0, where=(wtp_sorted >= flat_price_target),
                     color='#2ecc71', alpha=0.4, label='Lockean Flat Model Revenue')
    plt.plot(wtp_sorted, color='#2c3e50', linewidth=2.5, label='Individual Willingness-to-Pay (WTP)')
    plt.axhline(y=flat_price_target, color='#e74c3c', linestyle='--', linewidth=2, label='Lockean Flat Price ($350.00)')

    plt.title('Algorithmic Price Discrimination vs. The Lockean Common Measure', fontsize=12, fontweight='bold', pad=15)
    plt.xlabel('Synthetic Consumers (Sorted by Purchasing Power)', fontsize=10)
    plt.ylabel('Price / Budget ($)', fontsize=10)
    plt.legend(loc='upper right')
    plt.grid(axis='y', linestyle=':', alpha=0.6)
    plt.tight_layout()

    # Save chart as image to local workspace
    plt.savefig('wealth_extraction_simulation.png', dpi=300)
    print("✓ Chart visualization successfully saved as 'wealth_extraction_simulation.png'")
    plt.show()
```
